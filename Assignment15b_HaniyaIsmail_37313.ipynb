```python
df["snack_choice"].value_counts().plot(kind="bar")

plt.title("Distribution of Snack Choices")
plt.xlabel("Snack Choices")
plt.ylabel("Number of Students")

plt.show()
```

```python
pd.crosstab(df["age_group"], df["tea_type"]).plot(kind="bar")

plt.title("Tea Preference by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Students")

plt.show()
```
